## Data Analysis: 
*(Describe how you analyzed the data. This is where you should intersperse your Python code so that anyone reading this can run your code to perform the analysis that you did, generate your figures, etc.)*

To analyze the data, we wrote code to create a bar graph comparing Amyloid Beta-42 concentrations in males and females with dementia. Alongside this, code to conduct a Student's T-test was written for statistical comparison to observe if there was statistical difference between the two groups. Another bar graph was made to compare Amyloid Beta-42 concentration in patients of varying years of education. To test for statistical difference in these groups, an ANOVA test was conducted on the data. A scatter plot was also coded to show the correlation between Amyloid Beta-42 and age at death. An equation and R^2 value was evaluated for statistical analysis on this graph.

Before all of this could happen, we had to import libraries to call methods on. We also imported the data set to allow for objects to be made automatically instead of manually typing them in. The class method instantiate_from_csv was called on the Patient class to create each patient object directly from the .csv data set.

```python
from patient_Tess import *

# Imports necessary libraries
import matplotlib.pyplot as plt
from scipy import stats
import numpy as np
import statistics
import pandas as pd
from sklearn.linear_model import LinearRegression

# Creates patient objects directly from the .csv file without having to manually type them in
Patient.instantiate_from_csv("C:/Users/tmvit/OneDrive/Desktop/BME 2315/Module 1/BME2315_Module1/Metadata and Protein Data for Module 1.csv")
```

Below, a bar graph comparing Amyloid Beta-42 concentration in females and males with dementia is created.

```python
# Lists for bar graph: one for AB-42 concentration in females with dementia and one for males with dementia
AB42female = []
AB42male = []

# Filling bar graph lists with data from dataset: evaluating AB-42 concentration in females and males with dementia
# AB-42 levels for each gender go in separate lists to analyze later
for patient in Patient.filter(Patient.allPatients, sex = "Female", cognitiveStatus = "Dementia"):
    AB42female.append(patient.aBeta42)
for patient in Patient.filter(Patient.allPatients, sex = "Male", cognitiveStatus = "Dementia"):
    AB42male.append(patient.aBeta42)

# Assigns the mean of the AB-42 concentration for both male and female patients to separate variables which will be visible on the bar graph
x_AB42barF = (statistics.mean(AB42female))
x_AB42barM = (statistics.mean(AB42male))

# Assigns the standard deviation of the AB-42 concentration for both the male and female patients to separate variables
AB42barF_stdev = (statistics.stdev(AB42female))
AB42barM_stdev = (statistics.stdev(AB42male))

# Prints the mean and standard deviation of both the female and male groups for the graph
print(f'x_AB42barF = {x_AB42barF}, AB42barF_stdev {AB42barF_stdev}')
print(f'x_AB42barM = {x_AB42barM}, AB42barM_stdev {AB42barM_stdev}')

patientSex_cols = ['Female', 'Male'] # Defines the columns as female and male
mean_AB42 = [x_AB42barF, x_AB42barM] # Calculates the mean between the two variables
std_AB42 = [AB42barF_stdev, AB42barM_stdev] # Calculates the standard deviation between the two variables
yerr = [np.zeros(len(mean_AB42)), std_AB42] # Creates vertical error values

# Code below creates T-test statistic using the female and male AB42 concentrations
t_stat, p_val = stats.ttest_ind(AB42female, AB42male)
print(f't_stat = {t_stat}, p_val = {p_val}')

# Plots the bar graph using the columns and mean AB-42 previously defined in blue and orange color
plt.bar(patientSex_cols, mean_AB42, yerr=yerr, capsize=10, color=["blue", "orange"])
plt.title("Amyloid Beta-42 Concentrations in Female vs Male Patients with Dementia") # Main title of the graph
plt.xlabel("Sex") # X-axis label on graph
plt.ylabel("ABeta-42 (pg/ug)") # Y-axis label on graph
# Below is the code for the text on the graph
plt.text(
        0.5, 350, # This establishes where the code will be on the graph: 0.5 on x-axis (middle) and 350 on y-axis
        f"t = {t_stat:.2f}\np = {p_val:.3e}", # This brings the statistics and tells the system to round to 2 or 3 decimals
        ha = 'center', # Centers text horizontally at the x value
        va = 'bottom' # Centers bottom of text vertically at the y value
        )
plt.show() # Creates bar graph
```

This bar graph compares Amyloid Beta-42 concentrations in female versus male patients with dementia. The t-value and p-value were calculated and outlined on the graph. The p-value was calculated to be 0.4577, which is greater than 0.05, meaning we fail to reject the null hypothesis and do not confirm statistical difference between the two groups. This suggests the variability in the data is not explained by signficance between the groups; rather, it is due to chance or other variables. So, while the graph appears to explain that females with dementia have higher AB-42 concentrations than males with dementia, we cannot confirm with confidence that this is true.

Below, a bar graph comparing Amyloid Beta-42 concentrations in patients of varying years of education is created.

```python
# Filling bar graph lists with data from dataset: evaluating AB-42 concentration and years of education in patients
# AB-42 levels and years of education go into separate lists to analyze later
education_data = pd.DataFrame({ # DataFrame enters this data into a table
    "Years of Education": [patient.yearsOfEducation for patient in Patient.allPatients],
    "ABeta42": [patient.aBeta42 for patient in Patient.allPatients]
})

# Creates education_groups list for use in ANOVA test
education_groups = [
    group["ABeta42"] # Enters AB-42 concentrations into list
    for _, group in education_data.groupby("Years of Education") # Iterates through list and groups data by years of education
]

# Code below generates the f-value and p-value
f_stat, p_value = stats.f_oneway(*education_groups) # Calculates f and p values
print("F-statistic:", f_stat) # Prints f-value
print("p-value:", p_value) # Prints p-value
plt.text(5, 170, f"One-Way-ANOVA: p = {p_value:.3f}", # Plots p-value on the bar graph at position 5, 170 with placement in the top right
ha='right', va='top', fontsize=12)

education_summary = education_data.groupby("Years of Education")["ABeta42"].agg( # Groups data by years of education and then looks at the AB_42 concentration
    mean="mean", # Calculates mean AB-42 concentration for each year group
    std="std" # Calculates standard deviation of AB-42 for each year group
)

# Creates bar graph with years of education as the categories with corresponding mean AB-42 concentration
plt.bar(
    education_summary.index.astype(str),
    education_summary["mean"] 
)
plt.xlabel("Years of Education") # X-axis label on graph
plt.ylabel("Amyloid Beta-42 (pg/ug)") # X-axis label on graph
plt.title("Amyloid Beta-42 Concentration by Years of Education") # Main title of graph
plt.xticks(rotation=45) # Rotates the x-values to prevent crowding
plt.tight_layout() # Prevents overlapping while ensuring everything fits inside plot space
plt.show() # Creates bar graph
```

This bar graph compares years of eduation to the concentration of Amyloid Beta-42 in measured patients. An ANOVA test was run on the data to determine if there was statistical difference. It resulted in a p-value of 0.563, which is greater than 0.05, meaning we fail to reject the null hypothesis. This indicates there is no statistical difference between different years of education and AB-42 concentrations. Because the p-value is greater than 0.05, it is likely that the variance between the groups is due to chance rather than a correlation between the years of education and AB-42 concentration.

Below, a scatter plot comparing patient age at death with Amyloid Beta-42 concentration is created.

```python
# Lists for scatter plot: one for patient age at death and one for Amyloid Beta-42 concentration
patientAge = []
patientAB42 = []

# Filling scatter plot lists with data from dataset to analyze age at death vs AB-42 concentration
# Patient age at death gets its own list and so does AB-42 concentration to analyze in the scatter plot
for patient in Patient.allPatients:
    patientAge.append(patient.age)
for patient in Patient.allPatients:
    patientAB42.append(patient.aBeta42)

# X variable becomes the data stored in patient age at death, and y variable becomes the data stored in AB42 concentration
X = [patientAge]  # Independent variable
y = [patientAB42]   # Dependent variable

# Below reshapes the X and y values into an array for analyzing the slope and intercept of the line of best fit.
X = np.array(patientAge).reshape(-1,1)
y = np.array(patientAB42)

# Establishes the Linear Regression model fit
model = LinearRegression()
model.fit(X,y)

# The slope, intercept, and r2 value variables are defined and assigned their values
slope = model.coef_[0]
intercept = model.intercept_
r2 = model.score(X, y)

# The equation is formed and then plotted on the scatter plot in the top right corner
equation = f"y = {slope:.2f}x + {intercept:.2f}\nR² = {r2:.2f}"
plt.text(X.max()-15, y.max(), equation, color="red", fontsize=12, verticalalignment='top')

# Plots the scatter plot with the previously defined x and y variables in the color blue
plt.scatter(X, y, color='blue')
plt.plot(X, model.predict(X), color="red")
plt.xlabel('Age at Death') # X axis label
plt.ylabel('Amyloid Beta-42 (pg/ug)') # Y axis label
plt.title('Scatter Plot of Age at Death vs Amyloid Beta-42 Concentration') # Main title of scatter plot
plt.plot(X, model.predict(X), color="red")
plt.show() # Prints the scatter plot
```

This scatter plot shows the correlation between age at death and Amyloid Beta-42 concentration. The line of best fit shows that there is a slight negative correlation between age at death and AB-42 concentration. Though, the R^2 value of 0.02 suggests that only 2% of the variance in the age at death is explained by Amyloid Beta-42 concentration. This means that there is a very weak relationship between these two variables; either more variables are involved in the relationship, or AB-42 does not have a strong effect on a person's age at death.
