* **Prevalence & incidence:** It is estimated that 7.4 million Americans who are 65 years or older have Alzheimer’s Dementia, making the prevalence about 11% (1/9). About 200,000 Americans have younger onset dementia, with a prevalence of (0.11%) 110/100000 in 30 to 64 year olds. In 2011, the incidence of Alzheimer’s in people aged 65-74 was 0.4%, those aged 75-84 was 3.2%, and those 85 and older was 7.6%. It is believed the incidence is higher now in 2026.
https://www.alz.org/getmedia/ef8f48f9-ad36-48ea-87f9-b74034635c1e/alzheimers-facts-and-figures.pdf


* **Economic burden:**


* **Risk factors (genetic, lifestyle):** Age is the largest risk factor. Most people with dementia are 65 or older; people can still get dementia earlier, and this is called early onset dementia. This is much less common though. Another risk factor is genetics. APO-E is a gene that deals with the trafficking of cholesterol. There are three different variants: e2 (protective), e3 (neutral/most common), and e4 (increased Alzheimer’s Disease risk, pro inflammatory signaling). If you have 2 e3 alleles, you have low risk; if you have 1 e2 and 1 e4, then you have moderate risk; if you have 1 e3 and 1 e4 allele, then you have higher risk; if you have 2 e4 alleles, you have the highest risk.
https://www.alz.org/getmedia/ef8f48f9-ad36-48ea-87f9-b74034635c1e/alzheimers-facts-and-figures.pdf


* **Societal determinants:**


* **Symptoms:** Some symptoms include a difficulty remembering recent events or names, apathy, and depression. Later, communication problems, poor judgement, and behavioral changes occur. Lastly, a difficulty walking, speaking, and swallowing occur at the end of the disease. 
https://www.alz.org/getmedia/ef8f48f9-ad36-48ea-87f9-b74034635c1e/alzheimers-facts-and-figures.pdf


* **Diagnosis:**


* **Standard of care treatments (& reimbursement):** Medicare now covers certain FDA-approved monoclonal antibody treatments (these target amyloid) as long as clinicians enter data into a registry and they are approved. They also recently started covering brain amyloid positron emission tomography (PET) imaging Alzheimer's diagnosis. Hospital stays, doctors visits, and other prescription medicines are covered. Some medications exist for the cognitive and behavioral symptoms. Other than that, care from family members or facilities is important.
https://www.alz.org/getmedia/ef8f48f9-ad36-48ea-87f9-b74034635c1e/alzheimers-facts-and-figures.pdf
https://www.alz.org/alzheimers-dementia/treatments/cms-medicare-coverage


* **Disease progression & prognosis:**


* **Continuum of care providers:** Family members often recognize the first signs of dementia, and then primary care providers will diagnose and potentially provide referrals if needed. The next step is often a neurologist who can definitively diagnose and monitor the disease; they also work with the PCP and prescribe medications. Nurse practitioners and PA’s may step in to support the family and patient through education about Alzheimer’s. Occupational, speech, or physical therapists may also be consulted to address the decline and work to maintain independence. At a certain point, in-home care providers might become necessary to deliver care at home and support safety while maintaining daily activities. Support groups can also helped family members during this process.
https://www.alzheimers.gov/professionals/health-care-providers 
https://www.alz.org/getmedia/ef8f48f9-ad36-48ea-87f9-b74034635c1e/alzheimers-facts-and-figures.pdf

* **Clinical Trials/next-gen therapies:** There are currently 63 pharmacological trials (11 targeting amyloid, 7 targeting inflammation, 5 targeting receptors, 4 targeting tau, and many others targeting different areas). 50 of these are in phases I and II, and 13 are in phases III and IV. There are 148 non-pharmacological trials (39 about exercise, 24 about neurostimulation, 18 about cognitive training, 17 about sleep, and many other areas). 203 trials are testing dementia care and caregiving, with 44 of those working to improve caregiver health and well-being.
https://www.nia.nih.gov/research/ongoing-AD-trials
